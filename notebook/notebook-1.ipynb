```python
"""Lesson 1.4: Build a prompt from user input."""

# Collect details from the user
role = input("AI role (e.g. engineer): ")
topic = input("Topic: ")
level = input("Level (beginner/expert): ")

# Build the final prompt (this is what we will send to an LLM later)
prompt = f"You are a {role}. Explain {topic} to a {level} in 3 bullet points."

print("\n--- Your Prompt ---")
print(prompt)
```

```python
bakery = {
    "name": "Sweet Crumbs",
    "city": "Lahore",
    "hours": "9 am to 10 pm",
    "delivery": True,
}

print(bakery["name"])                      # Sweet Crumbs
print(bakery.get("phone", "not listed"))   # not listed (safe, no error)
print("city" in bakery)                    # True  (in checks the keys)
print(len(bakery))                         # 4
```

```python
business = {
    "name": "Sweet Crumbs",
    "prices": {"chocolate cake": 1500, "cupcake": 250, "brownie": 300},
    "best_sellers": ["chocolate cake", "brownie"],
}

print(business["prices"]["cupcake"])      # 250
print(business["best_sellers"][0])        # chocolate cake
business["prices"]["donut"] = 200         # add to the inner dictionary
print(len(business["prices"]))            # 4
```
